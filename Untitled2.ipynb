```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from textblob import TextBlob
from scipy.stats import spearmanr

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, accuracy_score, classification_report


df = pd.read_csv("Combined_News_DJIA.csv")


df["Date"] = pd.to_datetime(df["Date"])

# Combine Top1-Top25 news headlines into one text column
headline_columns = [col for col in df.columns if col.startswith("Top")]
df["combined_news"] = df[headline_columns].apply(
    lambda row: " ".join(row.astype(str)), axis=1
)

df = df[["Date", "Label", "combined_news"]]


df["sentiment"] = df["combined_news"].apply(
    lambda text: TextBlob(text).sentiment.polarity
)

df["headline_length"] = df["combined_news"].apply(len)
df["word_count"] = df["combined_news"].apply(lambda x: len(x.split()))


print("\n--- DATA INFO ---")
print(df.info())

print("\n--- DATA DESCRIPTION ---")
print(df.describe())

print("\n--- FIRST ROWS ---")
print(df.head())


# EDA 1: Market Movement Distribution

plt.figure(figsize=(6, 4))
sns.countplot(x=df["Label"])
plt.title("Market Up (1) vs Down (0)")
plt.xlabel("Market Movement")
plt.ylabel("Count")
plt.tight_layout()
plt.show()


# EDA 2: Monthly Market Trend

monthly_trend = df.set_index("Date")["Label"].resample("ME").mean()

plt.figure(figsize=(12, 5))
monthly_trend.plot()
plt.title("Monthly Average Market Movement")
plt.xlabel("Date")
plt.ylabel("Average Label")
plt.tight_layout()
plt.show()


# EDA 3: Sentiment Distribution

plt.figure(figsize=(8, 5))
sns.histplot(df["sentiment"], bins=30, kde=True)
plt.title("Distribution of News Sentiment Scores")
plt.xlabel("Sentiment Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# EDA 4: Sentiment vs Market Movement

plt.figure(figsize=(7, 5))
sns.boxplot(x=df["Label"], y=df["sentiment"])
plt.title("Sentiment Scores by Market Movement")
plt.xlabel("Market Movement (0 = Down, 1 = Up)")
plt.ylabel("Sentiment Score")
plt.tight_layout()
plt.show()


# HYPOTHESIS TEST

print("\n--- HYPOTHESIS TEST: Sentiment vs Market Movement ---")
print("H0: There is no relationship between news sentiment and market movement.")
print("H1: There is a relationship between news sentiment and market movement.")

corr, p_value = spearmanr(df["sentiment"], df["Label"])

print("Spearman Correlation:", corr)
print("p-value:", p_value)

if p_value < 0.05:
    print("Result: Reject H0 -> Sentiment is significantly related to market movement.")
else:
    print("Result: Fail to reject H0 -> No statistically significant relationship found.")


# TRAIN / TEST SPLIT
# According to dataset recommendation:
# Train: 2008-08-08 to 2014-12-31
# Test: 2015-01-02 to 2016-07-01

train = df[df["Date"] < "2015-01-01"]
test = df[df["Date"] >= "2015-01-01"]

features = ["sentiment", "headline_length", "word_count"]

X_train = train[features]
y_train = train["Label"]

X_test = test[features]
y_test = test["Label"]

print("\n--- TRAIN / TEST SIZE ---")
print("Train size:", len(train))
print("Test size:", len(test))


tree_model = DecisionTreeClassifier(random_state=42, max_depth=4)
tree_model.fit(X_train, y_train)

tree_pred = tree_model.predict(X_test)
tree_prob = tree_model.predict_proba(X_test)[:, 1]

tree_auc = roc_auc_score(y_test, tree_prob)
tree_acc = accuracy_score(y_test, tree_pred)

print("\n--- DECISION TREE PERFORMANCE ---")
print("Accuracy:", tree_acc)
print("AUC Score:", tree_auc)
print(classification_report(y_test, tree_pred))


rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    max_depth=5
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

rf_auc = roc_auc_score(y_test, rf_prob)
rf_acc = accuracy_score(y_test, rf_pred)

print("\n--- RANDOM FOREST PERFORMANCE ---")
print("Accuracy:", rf_acc)
print("AUC Score:", rf_auc)
print(classification_report(y_test, rf_pred))


importance = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

plt.figure(figsize=(7, 4))
sns.barplot(data=importance, x="Importance", y="Feature")
plt.title("Feature Importance - Random Forest")
plt.tight_layout()
plt.show()

print("\n--- FEATURE IMPORTANCE ---")
print(importance)


# FINAL INTERPRETATION

print("\n--- FINAL INTERPRETATION ---")

print("""
This project analyzed whether daily news sentiment is related to DJIA market movement.
The sentiment score was extracted from the top 25 daily news headlines using TextBlob.
A Spearman correlation test was applied to test the statistical relationship between
sentiment and market direction.

The machine learning models used sentiment, headline length, and word count as input features.
Decision Tree and Random Forest models were compared using Accuracy and ROC-AUC score.

If the AUC score is close to 0.50, this suggests that the model performs close to random guessing.
This does not mean the project failed. Instead, it shows that market movement is complex and cannot
be explained only by simple headline sentiment. Future work could improve the model by using
more advanced NLP methods such as FinBERT, including financial news sources, and adding numerical
market indicators such as trading volume, volatility, and previous-day returns.
""")
```
