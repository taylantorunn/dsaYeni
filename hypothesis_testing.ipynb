```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv("final_dataset.csv")

df["Label"] = (df["Close"].shift(-1) > df["Close"]).astype(int)
df = df.dropna()

up = df[df["Label"] == 1]["sentiment"]
down = df[df["Label"] == 0]["sentiment"]

t_stat, p_value = ttest_ind(up, down)

print("p-value:", p_value)
```

```python
# T-Test for Headline Length
up_length = df[df["Label"] == 1]["headline_length"]
down_length = df[df["Label"] == 0]["headline_length"]

t_stat_len, p_value_len = ttest_ind(up_length, down_length)
print("T-Test p-value for Headline Length:", p_value_len)
```
